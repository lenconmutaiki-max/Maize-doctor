```python

# MAIZE DOCTOR - NEW
# By Lencon Mutaiki
print("Mhoro! Maize Doctor")

s = input("What do you see? brown/yellow/white/black: ")
s = s.lower()

if "brown" in s:
    print("Fall Armyworm")
    print("Spray Lambda in morning")
elif "yellow" in s:
    print("Nitrogen Deficiency")
    print("Apply Ammonium Nitrate")
elif "white" in s:
    print("Maize Streak Virus")
    print("Remove infected plants")
elif "black" in s:
    print("Maize Smut")
    print("Remove and burn cobs")
else:
    print("Send to Agritex")

print("Built by Lencon Mutaiki")
```

```python

# TOBACCO DOCTOR - By Lencon Mutaiki
print("Mhoro! Tobacco Doctor")
s = input("What do you see? brown/yellow/black/white: ").lower()

if "brown" in s:
    print("Frog Eye Leaf Spot")
    print("Spray Copper Oxychloride, remove infected leaves")
elif "yellow" in s:
    print("Tobacco Mosaic Virus")
    print("Remove infected plants, wash hands/tools, no smoking in field")
elif "black" in s:
    print("Black Shank / Bacterial Wilt")
    print("Remove plant, rotate crop, do not plant tobacco there next year")
elif "white" in s:
    print("Powdery Mildew / White Mould")
    print("Spray Sulphur, improve air flow, remove lower leaves")
else:
    print("Send to Agritex / Tobacco Research Board")

print("Built by Lencon Mutaiki")
```
